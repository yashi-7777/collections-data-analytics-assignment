# Collections Analytics — Reproducible Analysis
This notebook reproduces the core audit, payment canonicalization and monthly recovery analysis.


```python
import pandas as pd, numpy as np
BASE='../collections'  # set to extracted raw dataset path
accounts=pd.read_csv(f'{BASE}/accounts.csv')
payments=pd.read_csv(f'{BASE}/payments.csv')
calls=pd.read_csv(f'{BASE}/calls.csv')
targeting=pd.read_csv(f'{BASE}/daily_targeting.csv')
```

```python
payments['event_at']=pd.to_datetime(payments['event_at'])
ranked=(payments.sort_values('event_at').groupby('payment_reference',as_index=False)
 .agg(account_id=('account_id','first'),amount=('amount','last'),status=('payment_status','last'),event_at=('event_at','last'),accounts=('account_id','nunique')))
canonical=ranked[(ranked.accounts==1)&(ranked.status=='SUCCESS')].copy()
canonical['month']=canonical.event_at.dt.to_period('M').astype(str)
canonical.groupby('month').amount.sum()
```

## Interpretation
Do not call the series a 12-month trend: most event data ends 8 Aug 2026. Compare raw SUCCESS to canonical recovery and document the gap.
